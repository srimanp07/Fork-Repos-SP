## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie

#Omschrijven van V=1/(1-(fv/fm)**2) naar fm=... en V=1/3 en w/w0 wordt f/f0 doordat de 2pi wegvallen. f=5 Hz
#Daar uit komt fm=root(25/4)
```

```python
# Voeg hier je plot in.
# Opdracht 2: 
# Plot: Bode -plot van het syteem met de gevonden eigenfrequentie: (het liefst aflezen op d ey-as is 1/3)
# Eigenfrequentie en bijberheonde deltax voor massa veer syteem gevonden
f_eigenfrequentie =  5/2
#f_vloer random aan het einde aanpassen aan wat daadwerkelijk is gedaan:
f_vloer = np.linspace(0.1, 100, 1000) #Hz
#f_vloer_echt = np.array([1,3,5,8]) #Hz

#overdracht formule: 
overdracht = np.abs(1 / (1 - (f_vloer/f_eigenfrequentie)**2))
#overdracht_echt = np.abs(1 / (1 - (f_vloer_echt/f_eigenfrequentie)**2))

plt.plot(f_vloer, overdracht)
#plt.plot(f_vloer_echt, overdracht_echt, 'ro') 

plt.xscale('log')
plt.yscale('log')

plt.title('Bode plot van de meetopstelling')
plt.xlabel('F_vloer (Hz)')
plt.ylabel('(A_out / A_in)')
plt.show()
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
dx=9.81/(2*np.pi*f_eigenfrequentie)**2
print(dx,'m')
```

```python
#simulatie voor boven minamle dx zitten> 

deltax=1.5*dx
fmassa=np.sqrt(9.81/deltax)/(2*np.pi)

#overdracht formule: 
overdracht = np.abs(1 / (1 - (f_vloer/fmassa)**2))
#overdracht_echt = np.abs(1 / (1 - (f_vloer_echt/f_eigenfrequentie)**2))

plt.plot(f_vloer, overdracht)
#plt.plot(f_vloer_echt, overdracht_echt, 'ro') 

plt.xscale('log')
plt.yscale('log')

plt.title('Bode plot van de meetopstelling')
plt.xlabel('F_vloer (Hz)')
plt.ylabel('(A_out / A_in)')
plt.show()
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().

# Plot van massa tegen veerconstante met de berekende deltax als constante en dan 3 grafieken om zoiezo boven minimale delta x waarde te zitten wegens onnaunwuekruoghedne (afweging nog maken)

g = 9.81  # Valversnelling

massa = np.linspace(0.1, 2.0, 100)

#Massa, veer formule uit eigenfruentie formule met krachtenbalans voor massa veer syteem zonder demping:
C1 = (massa * g) / dx          
C2 = (massa * g) / (2 * dx)    
C3 = (massa * g) / (5 * dx)   
C4 = (massa * g) / (1.5 * dx) 

plt.figure(figsize=(10, 6))
plt.plot(massa, C1, label = 'Minimale dx')
plt.plot(massa, C2, label = '2x dx')
plt.plot(massa, C3, label = '5x dx')
plt.plot(massa, C4, label = '1.5x dx')

plt.title('Veerconstante vs Massa')
plt.xlabel('M $kg$')
plt.ylabel('C $N/m$')
plt.grid(True)
plt.legend()

plt.show()
```
